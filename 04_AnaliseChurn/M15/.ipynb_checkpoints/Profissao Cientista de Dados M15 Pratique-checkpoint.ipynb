```python
df['churn'].value_counts().plot(kind='bar')
plt.title('Distribuição do Churn')
plt.xlabel('Churn')
plt.ylabel('Quantidade de Clientes')
plt.show()
```

Observa-se que a maior parte dos clientes não realizou churn, indicando um desbalanceamento da variável alvo, comportamento esperado em bases de retenção de clientes.

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['pagamento_mensal'])
plt.title('Boxplot do Pagamento Mensal')
plt.ylabel('Pagamento Mensal')
plt.show()
```

O boxplot do pagamento mensal indica uma distribuição relativamente assimétrica, com maior concentração de valores entre o primeiro e o terceiro quartil. Observa-se a presença de valores mais elevados, sugerindo possíveis outliers, o que confirma a necessidade de atenção a essa variável em análises posteriores.

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['total_pago'])
plt.title('Boxplot do Total Pago')
plt.ylabel('Total Pago')
plt.show()
```

O boxplot do total pago evidencia forte assimetria à direita, com alguns clientes apresentando valores acumulados muito superiores à maioria da base. Esse comportamento confirma a presença de outliers e reforça a heterogeneidade do tempo de relacionamento e consumo dos clientes.

```python
df['faixa_tempo_cliente'] = pd.cut(
    df['tempo_como_cliente'],
    bins=[0, 12, 24, 48, df['tempo_como_cliente'].max()],
    labels=['0-12 meses', '13-24 meses', '25-48 meses', '49+ meses'],
    include_lowest=True
)
```

```python
df['faixa_tempo_cliente'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribuição do Tempo como Cliente')
plt.ylabel('')
plt.show()
```

O gráfico de pizza mostra que a base de clientes está relativamente bem distribuída entre clientes novos (0–12 meses) e clientes com longo tempo de relacionamento (49+ meses), ambos representando cerca de um terço da base. Clientes com tempo intermediário representam uma parcela menor, indicando heterogeneidade no perfil de permanência dos clientes.

# 3) Identifique e trate as colunas que contém outliers.
Caso opte por mante-los ou altera-los justifique sua escolha.


```python
Q1 = df['total_pago'].quantile(0.25)
Q3 = df['total_pago'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

limite_inferior, limite_superior
```

```python
outliers_total_pago = df[
    (df['total_pago'] < limite_inferior) | 
    (df['total_pago'] > limite_superior)
]

outliers_total_pago.shape
```

```python
Q1 = df['pagamento_mensal'].quantile(0.25)
Q3 = df['pagamento_mensal'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

limite_inferior, limite_superior
```

Para a identificação de outliers, foi utilizado o método do Intervalo Interquartil (IQR) nas variáveis numéricas total pago e pagamento mensal. Após o cálculo dos limites inferior e superior, não foram identificados registros fora desses intervalos. Apesar da presença de valores elevados, estes se encontram dentro dos limites estatísticos esperados e representam comportamentos legítimos do negócio. Dessa forma, optou-se por manter todos os valores na base de dados, sem necessidade de tratamento adicional.

### O tipo de contrato influencia o churn?


```python
pd.crosstab(df['tipo_contrato'], df['churn']).plot(kind='bar')

plt.title('Distribuição do Churn por Tipo de Contrato')
plt.xlabel('Tipo de Contrato')
plt.ylabel('Quantidade de Clientes')
plt.legend(title='Churn')
plt.show()

```

Clientes com contrato mensal apresentam maior propensão ao churn, enquanto contratos de maior duração demonstram maior retenção.

### Pagamento Mensal por Churn

```python
df.boxplot(column='pagamento_mensal', by='churn')

plt.title('Pagamento Mensal por Churn')
plt.suptitle('')
plt.xlabel('Churn')
plt.ylabel('Pagamento Mensal')
plt.show()
```

Clientes que cancelaram tendem a pagar valores mensais mais altos do que aqueles que permaneceram na base.

### Tempo como cliente por Churn

```python
df.boxplot(column='tempo_como_cliente', by='churn')

plt.title('Tempo como Cliente por Churn')
plt.suptitle('')
plt.xlabel('Churn')
plt.ylabel('Tempo como Cliente (meses)')
plt.show()
```

Clientes com menor tempo como cliente tendem a apresentar maior probabilidade de churn, enquanto clientes mais antigos demonstram maior retenção.

### Churn por método de pagamento

```python
pd.crosstab(df['metodo_pagamento'], df['churn']).plot(kind='bar', figsize=(8,5))

plt.title('Distribuição do Churn por Método de Pagamento')
plt.xlabel('Método de Pagamento')
plt.ylabel('Quantidade de Clientes')
plt.legend(title='Churn')
plt.show()
```

Observa-se maior concentração de churn entre clientes que utilizam cheque eletrônico, enquanto métodos automáticos apresentam menor taxa de cancelamento.

### Churn por Dependentes

```python
pd.crosstab(df['possui_dependentes'], df['churn']).plot(kind='bar')

plt.title('Distribuição do Churn por Dependentes')
plt.xlabel('Possui Dependentes')
plt.ylabel('Quantidade de Clientes')
plt.legend(title='Churn')
plt.show()
```

Clientes com dependentes tendem a apresentar menor churn, indicando maior estabilidade desse perfil de clientes.

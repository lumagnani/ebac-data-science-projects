```python
while True:
    num1 = float(input('Adicione o primeiro número: '))
    num2 = float(input('Adicione o segundo número: '))

    print("\nEscolha uma operação:")
    print("1 - Soma (+)")
    print("2 - Subtração (-)")
    print("3 - Multiplicação (*)")
    print("4 - Divisão (/)")
    print("5 - Sair")

    opcao = input("Digite o número da operação: ")

    if opcao == '1':
        print(f"Resultado: {num1 + num2}")
    elif opcao == '2':
        print(f"Resultado: {num1 - num2}")
    elif opcao == '3':
        print(f"Resultado: {num1 * num2}")
    elif opcao == '4':
        if num2 != 0:
            print(f"Resultado: {num1 / num2}")
        else:
            print("Erro: Divisão por zero!")
    elif opcao == '5':
        print("Encerrando o programa.")
        break
    else:
        print("Opção inválida!")
5
```
